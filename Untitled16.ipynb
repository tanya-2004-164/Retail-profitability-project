Store-wise sales analysis

```python

query = """
SELECT
    Store_Name,
    ROUND(SUM(Sales_Revenue), 2) AS Total_Revenue,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    ROUND(
        100.0 * SUM(Profit) / SUM(Sales_Revenue), 2
    ) AS Profit_Margin_Percent
FROM retail_sales
GROUP BY Store_Name
ORDER BY Total_Profit DESC;
"""

pd.read_sql_query(query, conn)
```

```python

query = """
SELECT
    Category,
    ROUND(SUM(Sales_Revenue), 2) AS Total_Revenue,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    ROUND(
        100.0 * SUM(Profit) / SUM(Sales_Revenue), 2
    ) AS Profit_Margin_Percent
FROM retail_sales
GROUP BY Category
ORDER BY Total_Profit DESC;
"""

pd.read_sql_query(query, conn)
```

```python

query = """
SELECT
    CASE
        WHEN Discount_Pct = 0 THEN 'No Discount'
        WHEN Discount_Pct <= 0.10 THEN 'Low Discount (Up to 10%)'
        WHEN Discount_Pct <= 0.20 THEN 'Medium Discount (10-20%)'
        ELSE 'High Discount (Above 20%)'
    END AS Discount_Group,

    COUNT(*) AS Total_Orders,

    ROUND(SUM(Sales_Revenue), 2) AS Total_Revenue,

    ROUND(SUM(Profit), 2) AS Total_Profit,

    ROUND(
        100.0 * SUM(Profit) / SUM(Sales_Revenue), 2
    ) AS Profit_Margin_Percent

FROM retail_sales
GROUP BY Discount_Group
ORDER BY Profit_Margin_Percent DESC;
"""

pd.read_sql_query(query, conn)
```

Month-wise revenue and profit

```python

query = """
SELECT
    strftime('%Y-%m', Order_Date) AS Sales_Month,
    ROUND(SUM(Sales_Revenue), 2) AS Total_Revenue,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    COUNT(DISTINCT Order_ID) AS Total_Orders
FROM retail_sales
GROUP BY Sales_Month
ORDER BY Sales_Month;
"""

pd.read_sql_query(query, conn)
```

Negative profit order

```python

query = """
SELECT
    Order_ID,
    Order_Date,
    Store_Name,
    Product_Name,
    Sales_Revenue,
    Profit,
    Discount_Pct
FROM retail_sales
WHERE Profit < 0
ORDER BY Profit ASC
LIMIT 20;
"""

pd.read_sql_query(query, conn)
```

Final business insights

```python

query = """
SELECT
    'Overall Business' AS Analysis,
    ROUND(SUM(Sales_Revenue), 2) AS Revenue,
    ROUND(SUM(Profit), 2) AS Profit,
    ROUND(100.0 * SUM(Profit) / SUM(Sales_Revenue), 2)
        AS Profit_Margin_Percent
FROM retail_sales

UNION ALL

SELECT
    'Top Profitable Store',
    ROUND(SUM(Sales_Revenue), 2),
    ROUND(SUM(Profit), 2),
    ROUND(100.0 * SUM(Profit) / SUM(Sales_Revenue), 2)
FROM retail_sales
WHERE Store_Name = (
    SELECT Store_Name
    FROM retail_sales
    GROUP BY Store_Name
    ORDER BY SUM(Profit) DESC
    LIMIT 1
);
"""

pd.read_sql_query(query, conn)
```

```python

import pandas as pd
import plotly.express as px
from IPython.display import display, HTML

# Read data from the SQLite database you already created
df = pd.read_sql_query("SELECT * FROM retail_sales", conn)

# Prepare dates and month-wise data
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Sales_Month"] = df["Order_Date"].dt.to_period("M").astype(str)

# KPI calculations
revenue = df["Sales_Revenue"].sum()
profit = df["Profit"].sum()
margin = profit / revenue * 100

display(HTML(f"""
<div style="display:flex;gap:15px;flex-wrap:wrap;font-family:Arial">
  <div style="background:#e8f0fe;padding:20px;border-radius:12px;min-width:180px">
    <b>Total Revenue</b><h2>₹{revenue:,.0f}</h2>
  </div>
  <div style="background:#e6f4ea;padding:20px;border-radius:12px;min-width:180px">
    <b>Total Profit</b><h2>₹{profit:,.0f}</h2>
  </div>
  <div style="background:#fef7e0;padding:20px;border-radius:12px;min-width:180px">
    <b>Profit Margin</b><h2>{margin:.2f}%</h2>
  </div>
</div>
"""))

# Monthly sales trend
monthly = df.groupby("Sales_Month", as_index=False).agg(
    Revenue=("Sales_Revenue", "sum"),
    Profit=("Profit", "sum")
)
fig1 = px.line(
    monthly, x="Sales_Month", y=["Revenue", "Profit"],
    title="Monthly Revenue and Profit Trend", markers=True
)
fig1.show()

# Store-wise profit
stores = df.groupby("Store_Name", as_index=False)["Profit"].sum()
stores = stores.sort_values("Profit", ascending=False)
fig2 = px.bar(
    stores, x="Store_Name", y="Profit",
    title="Store-wise Profit", hover_data={"Profit":":,.2f"}
)
fig2.update_layout(xaxis_tickangle=-35)
fig2.show()

# Category-wise revenue
categories = df.groupby("Category", as_index=False).agg(
    Revenue=("Sales_Revenue", "sum"),
    Profit=("Profit", "sum")
)
fig3 = px.bar(
    categories, x="Category", y="Revenue",
    color="Profit", title="Category-wise Revenue and Profit",
    hover_data=["Profit"]
)
fig3.show()

# Product profitability
products = df.groupby("Product_Name", as_index=False).agg(
    Revenue=("Sales_Revenue", "sum"),
    Profit=("Profit", "sum")
)
products["Profit_Margin_Pct"] = products["Profit"] / products["Revenue"] * 100
products = products.sort_values("Profit", ascending=False)

fig4 = px.bar(
    products, x="Product_Name", y="Profit",
    title="Product-wise Profitability",
    hover_data={"Revenue":":,.2f", "Profit":":,.2f",
                "Profit_Margin_Pct":":.2f"}
)
fig4.update_layout(xaxis_tickangle=-45)
fig4.show()
```

```python

import plotly.graph_objects as go
from ipywidgets import interact, Dropdown
from IPython.display import display

def show_filtered_dashboard(region, category):
    filtered = df.copy()

    if region != "All":
        filtered = filtered[filtered["Region"] == region]

    if category != "All":
        filtered = filtered[filtered["Category"] == category]

    if filtered.empty:
        print("No data found for this selection.")
        return

    revenue = filtered["Sales_Revenue"].sum()
    profit = filtered["Profit"].sum()
    margin = profit / revenue * 100 if revenue else 0

    print(f"Total Revenue: ₹{revenue:,.2f}")
    print(f"Total Profit: ₹{profit:,.2f}")
    print(f"Profit Margin: {margin:.2f}%")
    print(f"Rows: {len(filtered)}")

    monthly = filtered.copy()
    monthly["Month"] = monthly["Order_Date"].dt.to_period("M").astype(str)
    monthly = monthly.groupby("Month", as_index=False)["Sales_Revenue"].sum()

    fig = px.line(
        monthly, x="Month", y="Sales_Revenue",
        markers=True, title="Filtered Monthly Revenue"
    )
    fig.show()

    cat = filtered.groupby("Category", as_index=False)["Profit"].sum()
    px.bar(
        cat, x="Category", y="Profit",
        title="Filtered Profit by Category"
    ).show()

interact(
    show_filtered_dashboard,
    region=Dropdown(
        options=["All"] + sorted(df["Region"].dropna().unique().tolist()),
        value="All", description="Region:"
    ),
    category=Dropdown(
        options=["All"] + sorted(df["Category"].dropna().unique().tolist()),
        value="All", description="Category:"
    )
);
```

Top 5 products by profit

```python

query = """
SELECT
    Product_Name,
    ROUND(SUM(Sales_Revenue), 2) AS Revenue,
    ROUND(SUM(Profit), 2) AS Profit,
    ROUND(
        100.0 * SUM(Profit) / SUM(Sales_Revenue), 2
    ) AS Profit_Margin_Percent
FROM retail_sales
GROUP BY Product_Name
ORDER BY SUM(Profit) DESC
LIMIT 5;
"""

top_products = pd.read_sql_query(query, conn)
display(top_products)

fig = px.bar(
    top_products.sort_values("Profit"),
    x="Profit",
    y="Product_Name",
    orientation="h",
    title="Top 5 Products by Total Profit",
    text="Profit"
)
fig.show()
```
